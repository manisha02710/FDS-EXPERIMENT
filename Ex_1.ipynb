```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Year": [2016, 2017, 2018, 2019, 2020],
    "Jobs": [250, 220, 350, 450, 550]
}

df = pd.DataFrame(data)

print(df)

plt.plot(df["Year"], df["Jobs"], marker="o")
plt.xlabel("Year")
plt.ylabel("Job Postings")
plt.title("Data Science Job Postings Trend")
plt.show()
```

```python
import pandas as pd

# Structured Data
data = {
    "Name": ["Ram", "Priya", "Arun"],
    "Age": [20, 21, 22],
    "Marks": [80, 90, 85]
}

df = pd.DataFrame(data)

print("STRUCTURED DATA")
print(df)

# Semi-Structured Data
print("\nSEMI-STRUCTURED DATA")

data = {
    "Name": "Ram",
    "Age": 20,
    "Skills": ["Python", "SQL"]
}

print(data)

# Unstructured Data
print("\nUNSTRUCTURED DATA")

data = [
    "Python is a programming language.",
    "Data Science deals with data.",
    "Machine Learning is part of AI."
]

for x in data:
    print(x)
```

```python
from cryptography.fernet import Fernet

# Generate encryption key
key = Fernet.generate_key()

# Create Fernet object
f = Fernet(key)

# Original message
message = b"Rajalakshmi Engineering College"

# Encrypt the message
token = f.encrypt(message)

print("Encrypted Data:")
print(token)

# Decrypt the message
original = f.decrypt(token)

print("\nDecrypted Data:")
print(original.decode())
```
