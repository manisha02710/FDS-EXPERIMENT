```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
sales = [20, 35, 30, 45, 50]

plt.plot(days, sales, marker="o")

plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Sales Data - Line Plot")
plt.grid(True)

plt.show()
```

```python
import matplotlib.pyplot as plt

products = ["Apple", "Banana", "Orange", "Mango", "Grapes"]
sales = [50, 35, 45, 60, 40]

plt.bar(products, sales)

plt.xlabel("Products")
plt.ylabel("Sales")
plt.title("Product Sales - Bar Chart")

plt.grid(axis='y')

plt.show()
```

```python
import matplotlib.pyplot as plt

products = ["Apple", "Banana", "Orange", "Mango", "Grapes"]
sales = [30, 20, 15, 25, 10]

plt.pie(
    sales,
    labels=products,
    autopct='%1.1f%%'
)

plt.title("Product Sales - Pie Chart")

plt.show()
```
