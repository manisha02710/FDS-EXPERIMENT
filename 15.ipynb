```python
import pandas as pd
import numpy as np

# New dataset
data = {
    "City": ["Chennai", "Mumbai", "Delhi", "Mumbai", "Bangalore",
             "Chennai", "Delhi", np.nan, "Bangalore", "Chennai"],
    "Age": [25, 32, 28, 35, 40, np.nan, 30, 45, 27, 38],
    "Income": [35000, 52000, 41000, np.nan, 68000,
               39000, 47000, 75000, 43000, 61000],
    "Selected": ["No", "Yes", "No", "Yes", "Yes",
                 "No", "Yes", "No", "Yes", "No"]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Handle missing values
df["City"] = df["City"].fillna(df["City"].mode()[0])
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Income"] = df["Income"].fillna(round(df["Income"].mean()))

# Convert City into numerical columns
city = pd.get_dummies(df["City"])

# Create preprocessed dataset
updated_dataset = pd.concat(
    [city, df[["Age", "Income", "Selected"]]],
    axis=1
)

# Convert Selected into 0 and 1
updated_dataset["Selected"] = updated_dataset["Selected"].replace(
    {"No": 0, "Yes": 1}
)

print("\nPreprocessed Dataset:")
print(updated_dataset)

print("\nMissing Values:")
print(updated_dataset.isnull().sum())
```
