```python
import pandas as pd


sample_data = {
    'ProductID': [101, 102, 103, 104, 105],
    'SaleDate': pd.to_datetime([
        '2023-01-01',
        '2023-01-02',
        '2023-01-03',
        '2023-01-04',
        '2023-01-05'
    ]),
    'Quantity': [10, 5, 8, 12, 7],
    'Price': [100.50, 200.75, 150.00, 50.25, 300.00]
}


df = pd.DataFrame(sample_data)


df.to_csv('sales_dataset.csv', index=False)


df = pd.read_csv('sales_dataset.csv')

print("SALES DATA")
print(df)

print("\nSUMMARY")
print(df.describe())

print("\nDUPLICATES")
print(df.duplicated().sum())
```
