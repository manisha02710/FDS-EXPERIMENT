```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler, MinMaxScaler

# New dataset
data = {
    "Study_Hours": [2, 4, 6, 8, 10],
    "Test_Score": [45, 55, 65, 80, 95],
    "Attendance": [60, 70, 75, 85, 95]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Standard Scaling
sc = StandardScaler()
standard_scaled = sc.fit_transform(df)

print("\nStandard Scaled Data:")
print(standard_scaled)

# Min-Max Scaling
mms = MinMaxScaler()
minmax_scaled = mms.fit_transform(df)

print("\nMin-Max Scaled Data:")
print(minmax_scaled)
```
