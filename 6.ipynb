```python
import pandas as pd
import matplotlib.pyplot as plt


data = {
    "Date": [
        "2026-01-01",
        "2026-01-02",
        "2026-01-03",
        "2026-01-04",
        "2026-01-05",
        "2026-01-06",
        "2026-01-07"
    ],
    "Visitors": [120, 150, 180, 160, 220, 250, 230]
}


df = pd.DataFrame(data)


df["Date"] = pd.to_datetime(df["Date"])


df.set_index("Date", inplace=True)


print("Website Traffic Data:")
print(df)


print("\nStatistical Summary:")
print(df.describe())


print("\nAverage Website Visitors:", df["Visitors"].mean())


print("Maximum Website Visitors:", df["Visitors"].max())


print("Minimum Website Visitors:", df["Visitors"].min())


plt.figure(figsize=(10, 5))
plt.plot(df.index, df["Visitors"], marker="o", color="blue")

plt.title("Time Series Analysis of Website Traffic")
plt.xlabel("Date")
plt.ylabel("Number of Visitors")
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```
